```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Датасетти чакырып алуу
df = pd.read_csv('../data/student_data.csv')

# Биринчи 5 сапты көрүү
df.head()
```

```python
# Размер таблицы (саптар жана колонкалар саны)
print("Размер датасета:", df.shape)

# Типы данных и пропуски
df.info()

# Базовые статистики
df.describe()
```

### Data Dictionary

| Признак | Тип данных | Описание |
| :--- | :--- | :--- |
| `school` | Categorical | Школа студента ('GP' - Gabriel Pereira или 'MS' - Mousinho da Silveira) |
| `sex` | Categorical | Пол студента ('F' - женский, 'M' - мужской) |
| `age` | Numeric (Integer) | Возраст студента (от 15 до 22) |
| `studytime` | Numeric (Integer) | Еженедельное время учебы (1 - <2 часов, 2 - 2-5 часов, 3 - 5-10 часов, 4 - >10 часов) |
| `G3` | Numeric (Integer) | Итоговая оценка (от 0 до 20) |


```python
# График 1: Распределение возраста
plt.figure(figsize=(8, 4))
sns.histplot(df['age'], kde=True, color='skyblue')
plt.title('Распределение возраста студентов')
plt.xlabel('Возраст')
plt.ylabel('Количество')
plt.show()

# График 2: Соотношение по полу
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='sex', palette='Set2')
plt.title('Соотношение студентов по полу')
plt.xlabel('Пол (F - женский, M - мужской)')
plt.ylabel('Количество')
plt.show()
```

### Основные наблюдения о данных
1. **Структура и целостность**: Датасет содержит сведения о студентах с числовыми и категориальными признаками, пропущенные значения отсутствуют.
2. **Возрастной состав**: Основная часть студентов находится в возрасте от 15 до 18 лет.
3. **Перспектива ML**: Признак `studytime` и промежуточные оценки подходят для прогнозирования итоговой успеваемости (`G3`) в рамках будущего семестрового проекта.
